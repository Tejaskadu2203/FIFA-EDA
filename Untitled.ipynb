```python
df['Weight_in_kgs']=df['Weight_in_kgs'].astype(int)
```

```python
df['Hits']=df['Hits'].apply(lambda x:float(x.replace('K',''))*1000 if 'K' in str(x) else float(x))
```

```python
# Mean will result in high due to much greater values so going for median will be more suitable

df['Hits'].median()
```

```python
# Replacing null values with median

df['Hits']=df['Hits'].fillna(df['Hits'].median())
```

```python
# Converting Float to Int  
df['Hits']=df['Hits'].astype(int)
```

```python
# Replacing with Mode for Categorical Value 
loan=df['Loan Date End'].mode()[0]

df['Loan Date End']=df['Loan Date End'].fillna(loan)
```

```python
# Converting datatype from str to datetime
df['Loan Ending Month']=df['Loan Date End'].str.split(' ').str[0]
df['Loan Ending Date']=df['Loan Date End'].str.split(' ').str[1].str.split(',').str[0]
df['Loan Ending Year']=df['Loan Date End'].str.split(',').str[1]
```

```python
# Dropping Unwanted Columns

df.drop('Loan Date End',axis=1,inplace=True)
```

```python
# Converting datatypes 
df['Loan Ending Date']=df['Loan Ending Date'].astype(int)
df['Loan Ending Year']=df['Loan Ending Year'].astype(int)
```

```python
# Plot 1
import matplotlib.pyplot as plt 
top5_hits=df.groupby('Hits')['Name'].max().tail().sort_values()

plt.figure(figsize=(9,5))
color=['red','blue','orange','pink','yellow']
plt.bar(top5_hits.values,top5_hits.index,color=color)
plt.xlabel('Name',fontweight='bold')
plt.ylabel('Hits',fontweight='bold')
plt.grid(axis='y')
plt.title('Top 5 Players with Highest Hits',fontweight='bold')
plt.show()
```

```python
# Club with most players
#Plot 2

# Horizontal Bar Chart
df['Club'].value_counts().sort_values().tail().plot(kind='barh',color='lightcoral')
plt.title('Club with Most Players',fontweight='bold')
```

```python
# Summary for players count with rating 

df['IR'].value_counts().sort_index()
```

```python
# Plot 3
import seaborn as sns 
rating = df['IR'].value_counts(normalize=True) * 100

sns.barplot(x=rating.index, y=rating.values,color='pink')
plt.ylabel("Percentage")
plt.title('Percentage of Players with Most Rating',fontweight='bold')
```

```python
# Plot 4
sns.histplot(df['Age'],bins=10,kde=True)
plt.title("No of Player's in Specific Age",fontweight='bold',fontsize=17)
```

```python
# Plot 5
df['Preferred Foot'].value_counts().plot(kind='pie',autopct='%1.1f%%',colors=['lightgreen','lightblue'],shadow='black')
plt.title('Players with Preferred Foot',fontweight='bold',fontstyle='italic')
plt.ylabel('Count',fontweight='bold')
```

```python
# Plot 6
color=['lightblue']
df[['Name','Sprint Speed']].nlargest(5,'Sprint Speed').set_index('Name').plot(kind='barh',color=color)
plt.title('Top 5 Players with Highest Sprint Speed',fontweight='bold')
plt.ylabel('Name',fontweight='bold')
plt.legend()
plt.show()
```

```python
# Plot 7
sns.scatterplot(x='Age',y='PHY',data=df)
plt.title('Age V/s Fitness',fontweight='bold',fontsize=14,fontstyle='italic')
plt.ylabel('PHY',fontweight='bold')
plt.xlabel('Age',fontweight='bold')
```
